# MiraBest Dataset Exploration

This notebook loads the official MiraBest MBFRConfident subset for the binary FR-I versus FR-II task. No preprocessing, augmentation, or model training is performed here.

```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.data.download_mirabest import load_mirabest

dataset = load_mirabest(PROJECT_ROOT / 'data' / 'raw' / 'mirabest', download=True)
```

## Dataset size

The value below is calculated from the loaded train and test files.

```python
print(f'Loaded samples: {len(dataset)}')
```

## Actual class counts

Labels are remapped by the official loader to `0` for FR-I and `1` for FR-II.

```python
from collections import Counter

label_names = {0: 'FR-I', 1: 'FR-II'}
counts = Counter(int(dataset[index][1]) for index in range(len(dataset)))
for label, name in label_names.items():
    print(f'{name}: {counts[label]}')
```

## Real sample images

The following grid displays loaded grayscale images without augmentation or artificial modification. Each title is the true binary label returned by the loader.

```python
samples = {0: [], 1: []}
for index in range(len(dataset)):
    image, label = dataset[index]
    if len(samples[label]) < 5:
        samples[label].append(image)
    if all(len(images) == 5 for images in samples.values()):
        break

figure, axes = plt.subplots(2, 5, figsize=(12, 5))
for row, label in enumerate((0, 1)):
    for column, image in enumerate(samples[label]):
        axes[row, column].imshow(image, cmap='gray')
        axes[row, column].set_title(label_names[label])
        axes[row, column].axis('off')
figure.tight_layout()
plt.show()
```
