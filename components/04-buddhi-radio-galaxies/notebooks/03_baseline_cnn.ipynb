# Phase 3: Baseline CNN

This notebook inspects the intentionally small baseline CNN for the MiraBest Zenodo v1 binary task. It accepts normalized `1x150x150` images and returns two class logits. Full training is started from `scripts/train_model.py`, not automatically from this notebook.

```python
import torch

from src.models.baseline_cnn import BaselineCNN

model = BaselineCNN()
model.eval()
example_batch = torch.zeros(2, 1, 150, 150)
with torch.no_grad():
    logits = model(example_batch)
print(model)
print('Input shape:', tuple(example_batch.shape))
print('Logit shape:', tuple(logits.shape))
```

Training uses CrossEntropyLoss, Adam, the saved train/validation split, early stopping on validation loss, and CPU-compatible settings from `config.yaml`. The test split is not used in this phase.
